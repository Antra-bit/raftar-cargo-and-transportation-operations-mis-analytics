# Transport Operations MIS & Performance Analytics System
## Notebook 03: Exploratory Data Analysis (EDA)
**Organization**: Raftaar Cargo & Transport Services Pvt. Ltd.  
**Objective**: Conduct statistical analysis on validated shipment telemetry, assess corridor volume distributions, identify delay drivers, and analyze route cost economics.

```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/processed/validated_shipments.csv')
df_expenses = pd.read_csv('../data/raw/expenses.csv')
df = pd.merge(df, df_expenses, on='Shipment_ID', how='left')
print(f"Libraries imported. Dataset loaded: {len(df)} validated shipments.")
```

### 1. Delivery Status Distribution
Analyzing the operational split across active logistics states.

```python
status_counts = df['Status'].value_counts()
status_pct = df['Status'].value_counts(normalize=True) * 100
print("Status Split:")
for s, count in status_counts.items():
    print(f"{s:<12} {count:>3} ({status_pct[s]:.2f}%)")
```

### 2. High-Volume Freight Corridors
Top origin-destination pairs by trip count and cumulative revenue.

```python
routes = df.groupby(['Origin', 'Destination']).agg({
    'Shipment_ID': 'count',
    'Freight': 'mean',
    'Distance_KM': 'mean'
}).rename(columns={'Shipment_ID': 'Trips'}).sort_values(by='Trips', ascending=False)

print(f"{'Route':<25} {'Trips':>5}   {'Avg Freight (INR)':>17}   {'Avg Dist (KM)':>13}")
for idx, row in routes.head(5).iterrows():
    route_name = f"{idx[0]} -> {idx[1]}"
    print(f"{route_name:<25} {int(row['Trips']):>5}   {row['Freight']:>17,.0f}   {row['Distance_KM']:>13,.0f}")
```
