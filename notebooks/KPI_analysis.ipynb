# Transport Operations MIS & Performance Analytics System
## Notebook 04: KPI Calculation & Management Information System (MIS) Engine
**Organization**: Raftaar Cargo & Transport Services Pvt. Ltd.  
**Objective**: Compute the 8 core operational KPIs, generate Daily, Weekly, and Monthly MIS executive summaries, and produce operational variance indicators.

```python
import pandas as pd
import numpy as np

df_shipments = pd.read_csv('../data/processed/validated_shipments.csv')
df_expenses = pd.read_csv('../data/raw/expenses.csv')
df = pd.merge(df_shipments, df_expenses, on='Shipment_ID', how='left').fillna(0)
df['Total_Cost'] = df['Fuel_Cost'] + df['Toll_Cost'] + df['Maintenance_Cost'] + df['Other_Cost']
print("Operational KPI Calculation Engine Loaded.")
```

### Core 8 Operations KPIs
Executing the standardized mathematical formulas.

```python
kpi_1_total = len(df)
kpi_2_delivered = (df['Status'] == 'Delivered').sum()
kpi_3_delayed = (df['Status'] == 'Delayed').sum()
completed = kpi_2_delivered + kpi_3_delayed
kpi_4_otd = round((kpi_2_delivered / completed) * 100, 2) if completed > 0 else 0.0

df['Booking_Date'] = pd.to_datetime(df['Booking_Date'])
df['Actual_Delivery'] = pd.to_datetime(df['Actual_Delivery'])
durations = (df['Actual_Delivery'] - df['Booking_Date']).dt.days.dropna()
kpi_5_avg_time = round(durations.mean(), 2)

kpi_6_cost = df['Total_Cost'].sum()
kpi_7_cost_per_shp = round(kpi_6_cost / kpi_1_total, 2)
kpi_8_cost_per_km = round(kpi_6_cost / df['Distance_KM'].sum(), 2)
total_revenue = df['Freight'].sum()
margin_pct = round(((total_revenue - kpi_6_cost) / total_revenue) * 100, 2)

print("="*61)
print("RAFTAR OPERATIONS EXECUTIVE KPI SCORECARD")
print("="*61)
print(f"KPI 1: Total Shipments Processed         : {kpi_1_total}")
print(f"KPI 2: Successfully Delivered Shipments   : {kpi_2_delivered}")
print(f"KPI 3: Delayed Shipments                 : {kpi_3_delayed}")
print(f"KPI 4: On-Time Delivery Rate (OTD %)     : {kpi_4_otd}%")
print(f"KPI 5: Average Delivery Time (Days)      : {kpi_5_avg_time} Days")
print(f"KPI 6: Total Operational Cost (INR)      : ₹ {kpi_6_cost:,.2f}")
print(f"KPI 7: Cost Per Shipment (INR)           : ₹ {kpi_7_cost_per_shp:,.2f}")
print(f"KPI 8: Cost Per KM (INR/KM)              : ₹ {kpi_8_cost_per_km:.2f} / KM")
print("-"*61)
print(f"Gross Freight Revenue                    : ₹ {total_revenue:,.2f}")
print(f"Net Operating Margin                     : {margin_pct}%")
print("="*61)
```
